```python
!pip install -q google-genai

import os
from google import genai

os.environ["GEMINI_API_KEY"] = "aidjbahdbahbdskabdkjsabd-dasjsd--adnjsan"

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
model = "gemini-3.5-flash-lite"

qa_text = """
Artificial Intelligence is a branch of computer science that focuses on creating
systems capable of performing tasks that normally require human intelligence.
These tasks include learning, reasoning, problem solving, understanding language,
and recognizing patterns.
"""

qa_question = "What are the main tasks performed by Artificial Intelligence?"

qa_expected = [
    "learning",
    "reasoning",
    "problem solving",
    "understanding language",
    "recognizing patterns"
]

summary_text = """
Artificial Intelligence has become an important technology in modern healthcare.
It is used for disease detection, medical image analysis, drug discovery,
patient monitoring, and personalized treatment. AI can analyze large amounts of
medical data quickly and help doctors make better decisions. However, challenges
such as data privacy, bias, and reliability must be carefully addressed.
"""

summary_expected = """
AI is used in healthcare for disease detection, medical image analysis,
drug discovery, patient monitoring, and personalized treatment. It helps
analyze medical data and support doctors, but privacy, bias, and reliability
remain important challenges.
"""

sentiment_data = [
    ("I love this product. It works perfectly and I am very happy.", "Positive"),
    ("The product is terrible and completely useless.", "Negative"),
    ("The product arrived yesterday and has several features.", "Neutral"),
    ("The service was excellent and the staff were very helpful.", "Positive"),
    ("I am disappointed with the poor quality of this product.", "Negative")
]

qa_prompt = f"""
Answer the following question using the given text.

Text:
{qa_text}

Question:
{qa_question}
"""

summary_prompt = f"""
Summarize the following text in 2-3 sentences.

Text:
{summary_text}
"""

qa_response = client.models.generate_content(
    model=model,
    contents=qa_prompt
)

summary_response = client.models.generate_content(
    model=model,
    contents=summary_prompt
)

print("=" * 60)
print("QUESTION ANSWERING")
print("=" * 60)
print(qa_response.text)

qa_score = sum(
    word.lower() in qa_response.text.lower()
    for word in qa_expected
) / len(qa_expected) * 100

print("\nQA Score:", round(qa_score, 2), "%")

print("\n" + "=" * 60)
print("TEXT SUMMARIZATION")
print("=" * 60)
print(summary_response.text)

summary_words = set(summary_response.text.lower().split())
expected_words = set(summary_expected.lower().split())

summary_score = len(summary_words & expected_words) / len(expected_words) * 100

print("\nSummary Similarity Score:", round(summary_score, 2), "%")

print("\n" + "=" * 60)
print("SENTIMENT ANALYSIS")
print("=" * 60)

correct = 0

for text, expected in sentiment_data:

    prompt = f"""
Classify the sentiment of the following text as exactly one of:
Positive, Negative, Neutral.

Text:
{text}

Return only the sentiment label.
"""

    response = client.models.generate_content(
        model=model,
        contents=prompt
    )

    predicted = response.text.strip()

    if expected.lower() in predicted.lower():
        correct += 1

    print("\nText:", text)
    print("Expected:", expected)
    print("Predicted:", predicted)

sentiment_accuracy = correct / len(sentiment_data) * 100

print("\nSentiment Accuracy:", round(sentiment_accuracy, 2), "%")

print("\n" + "=" * 60)
print("FINAL EVALUATION")
print("=" * 60)

print("Question Answering Score:", round(qa_score, 2), "%")
print("Summarization Similarity:", round(summary_score, 2), "%")
print("Sentiment Accuracy:", round(sentiment_accuracy, 2), "%")
```
